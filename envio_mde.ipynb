Consulta de eventos de já manifestados

```python
import requests
import base64
import pandas as pd

# URL da API
url = "https://managersaashom.tecnospeed.com.br:7071/ManagerAPIWeb/mdfe/consulta"
df = pd.read_excel('/content/drive/MyDrive/MDe/layout_mde.xlsx')

# usuário e senha do Manager SaaS
usuario = "admin"
senha = ""

credenciais = f"{usuario}:{senha}"
token = base64.b64encode(credenciais.encode()).decode()

for index, row in df.iterrows():
  cnpj_emitente = str(row['cnpj_emitente'])
  chave_nota = str(row['chave_nota']).strip()
  tipo_evento = str(row['tipo_evento'])

  headers = {
      "Authorization": f"Basic {token}"
  }

  # parâmetros da consulta
  params = {
      "Grupo": "PLUG_24217653000195",
      "CNPJ": cnpj_emitente,
      "Filtro": f"chave = {chave_nota}",
      "Campos": "chave,situacao,nprotenvio,motivo",
      "Limite": "100"
  }

  response = requests.get(url, headers=headers, params=params)

  print("Status:", response.status_code)
  print("Resposta:", response.text)
```

Manifestação Documento fiscal

```python
import requests
import pandas as pd
import base64
from datetime import datetime

url = "https://managersaashom.tecnospeed.com.br:7071/ManagerAPIWeb/nfe/envia"

# usuário e senha do Manager SaaS
usuario = "paulo.lomanto@loggi.com"
senha = "D3A8475D2B0A5BE8017DBF9F6C812707"

# GERA TOKEN BASE64
credenciais = f"{usuario}:{senha}"
token = base64.b64encode(credenciais.encode()).decode()

# HEADERS
headers = {
    "Content-Type": "application/x-www-form-urlencoded",
    "Authorization": f"Basic {token}"
}

grupo_emitente = "loggi-mdfe"

df = pd.read_excel('/content/drive/MyDrive/MDe/layout_mde.xlsx')
data_hora_evento = datetime.now().strftime("%Y-%m-%dT%H:%M:%S")

for index, row in df.iterrows():

    cnpj_emitente = str(row['cnpj_emitente'])
    chave_nota = str(row['chave_nota']).strip()
    tipo_evento = str(row['tipo_evento'])

    string_arquivo = (
        f"DOCUMENTO=MDE "
        f"TIPOEVENTO={tipo_evento} "
        f"CHAVENOTA={chave_nota} "
        f"DHEVENTO={data_hora_evento} "
        f"FUSO=-03:00 "
        f"SEQEVENTO=1"
    )

    payload = {
        "encode": "true",
        "grupo": grupo_emitente,
        "cnpj": cnpj_emitente,
        "arquivo": string_arquivo
    }

    print(f"Enviando manifestação para a chave: {chave_nota}")

    response = requests.post(
        url,
        headers=headers,
        data=payload
    )

    print(f"Status Code: {response.status_code}")
    print(f"Resposta: {response.text}\n")
```
